```python
"""
====================================================================
 LABIRINTO DE ÁRVORES  —  jogo de labirinto usando árvore, em Python
====================================================================

Regras do exercício:
  - o labirinto é modelado como uma árvore;
  - cada nível tem 5 opções possíveis;
  - a quantidade de níveis = quantidade de letras do primeiro nome
    de quem joga;
  - a única sequência correta de opções é sorteada com random no
    início do jogo;
  - a cada escolha o jogo avisa se foi certa ou errada;
  - BÔNUS: a cada tentativa, a árvore explorada é impressa até o
    nível máximo já alcançado;
  - o número de tentativas até concluir o labirinto é registrado;
  - ao concluir, o jogo imprime a frase "parabens chefia tu passou".

Como pedido no exercício, os nomes de variáveis/funções/classes usam
músicas do Arlindo Cruz. Legenda (música -> o que ela representa aqui):

    MeuLugar                       -> classe do nó da árvore (uma "casa"
                                       do labirinto)
    o_que_e_o_amor                  -> nome de quem joga (fixo: "Vinicius"),
                                       define a quantidade de níveis
    a_pureza_da_flor                -> quantidade de níveis do labirinto
    to_a_bangu                      -> nó raiz da árvore (ponto de partida)
    ainda_e_tempo_pra_ser_feliz()   -> monta a árvore e sorteia o único
                                       caminho correto
    sera_que_e_amor                 -> opção certa (1 a 5) sorteada em
                                       cada nível
    fim_da_tristeza                 -> próximo nó da árvore, criado a
                                       cada nível
    bagaco_da_laranja()              -> [BÔNUS] imprime a árvore até o
                                       nível máximo já alcançado
    saudade_louca                   -> contador de tentativas
    camarao_que_dorme_a_onda_leva   -> nível atual dentro da tentativa
    agora_viu_que_perdeu_e_chora    -> True quando o jogador erra e a
                                       tentativa acaba
    o_show_tem_que_continuar        -> True enquanto o labirinto inteiro
                                       ainda não foi concluído
    samba_de_arere                  -> maior nível já alcançado (recorde)
    a_sos                           -> opção (1 a 5) escolhida pelo jogador
    o_bem                           -> marca a opção certa ao imprimir a
                                       árvore
====================================================================
"""

import sys
import random

# Evita erro de acentuação/emoji em consoles Windows mais antigos.
try:
    sys.stdout.reconfigure(encoding="utf-8")
except Exception:
    pass

# Nome de quem está resolvendo o exercício - define a quantidade de níveis.
o_que_e_o_amor = "Vinicius"


class MeuLugar:
    """Nó da árvore: representa uma 'casa' (checkpoint) do labirinto."""

    def __init__(self, nivel):
        self.nivel = nivel
        self.opcoes = {}              # {1..5: próximo MeuLugar ou None}
        self.sera_que_e_amor = None   # opção certa (1-5) NESTE nó


def ainda_e_tempo_pra_ser_feliz(a_pureza_da_flor):
    """
    Monta a árvore do labirinto (com 'a_pureza_da_flor' níveis, 5 opções
    em cada) e sorteia, usando random, a única sequência de opções que
    leva do início ao fim do labirinto.
    """
    to_a_bangu = MeuLugar(0)      # raiz = ponto de partida do labirinto
    no_atual = to_a_bangu

    for nivel in range(1, a_pureza_da_flor + 1):
        sera_que_e_amor = random.randint(1, 5)    # opção certa sorteada
        fim_da_tristeza = MeuLugar(nivel)          # próximo checkpoint

        no_atual.opcoes = {
            opcao: (fim_da_tristeza if opcao == sera_que_e_amor else None)
            for opcao in range(1, 6)
        }
        no_atual.sera_que_e_amor = sera_que_e_amor
        no_atual = fim_da_tristeza

    return to_a_bangu


def bagaco_da_laranja(to_a_bangu, samba_de_arere):
    """
    [BÔNUS] Imprime a árvore explorada até o maior nível já alcançado
    pelo jogador (samba_de_arere = recorde de nível alcançado).
    """
    print("\n🌳 Seu avanço no labirinto até agora:")

    if samba_de_arere == 0:
        print("   (nenhum nível conquistado ainda)\n")
        return

    no_atual = to_a_bangu
    for nivel in range(1, samba_de_arere + 1):
        recuo = "   " * (nivel - 1)
        o_bem = no_atual.sera_que_e_amor
        print(f"{recuo}Nível {nivel}")
        for opcao in range(1, 6):
            marca = "[OK] caminho certo" if opcao == o_bem else "[X]"
            print(f"{recuo}  └── opção {opcao}  {marca}")
        no_atual = no_atual.opcoes[o_bem]
    print()


def main():
    print("=" * 54)
    print("   🔀  BEM-VINDO AO LABIRINTO DE ÁRVORES  🔀")
    print("=" * 54)

    nome = o_que_e_o_amor
    a_pureza_da_flor = len(nome)
    print(f"\n{nome}, seu labirinto terá {a_pureza_da_flor} níveis "
          f"(1 nível para cada letra do seu nome). Boa sorte!")

    to_a_bangu = ainda_e_tempo_pra_ser_feliz(a_pureza_da_flor)

    saudade_louca = 0                  # contador de tentativas
    samba_de_arere = 0                 # recorde: maior nível já alcançado
    o_show_tem_que_continuar = True    # labirinto ainda não concluído

    while o_show_tem_que_continuar:
        saudade_louca += 1
        print(f"\n----- Tentativa {saudade_louca} -----")

        no_atual = to_a_bangu
        camarao_que_dorme_a_onda_leva = 0      # nível atual da tentativa
        agora_viu_que_perdeu_e_chora = False   # errou e a tentativa acabou?

        while (not agora_viu_que_perdeu_e_chora
               and camarao_que_dorme_a_onda_leva < a_pureza_da_flor):

            camarao_que_dorme_a_onda_leva += 1
            print(f"\nNível {camarao_que_dorme_a_onda_leva} de {a_pureza_da_flor}")

            a_sos = input("Escolha um caminho (1 a 5): ").strip()
            while a_sos not in ("1", "2", "3", "4", "5"):
                a_sos = input("Opção inválida! Escolha de 1 a 5: ").strip()
            a_sos = int(a_sos)

            if a_sos == no_atual.sera_que_e_amor:
                print("✅ Caminho certo! Pode seguir para o próximo nível.")
                no_atual = no_atual.opcoes[a_sos]
                samba_de_arere = max(samba_de_arere, camarao_que_dorme_a_onda_leva)
            else:
                print("❌ Caminho errado! Essa tentativa termina aqui.")
                agora_viu_que_perdeu_e_chora = True

        if not agora_viu_que_perdeu_e_chora:
            # chegou ao fim de todos os níveis sem nenhum erro
            o_show_tem_que_continuar = False

        bagaco_da_laranja(to_a_bangu, samba_de_arere)

    palavra_tentativa = "tentativa" if saudade_louca == 1 else "tentativas"

    print("🎉" * 27)
    print("parabens chefia tu passou")
    print("🎉" * 27)
    print(f"\n{nome}, você concluiu o labirinto em "
          f"{saudade_louca} {palavra_tentativa}!")


if __name__ == "__main__":
    main()
```
